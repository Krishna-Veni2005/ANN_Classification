```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import pickle
from tensorflow.keras.models import load_model
import tensorflow as tf
```

```python
## load the trained model,scaler pickle, onehot
model=load_model('model.h5')
with open('onehot_encoder_geo.pkl', 'rb') as file:
    onehot_encoder_geo = pickle.load(file)
with open('label_encoder_gender.pkl', 'rb') as file:
    label_encoder_gender = pickle.load(file)
with open('scaler.pkl','rb') as file:
    scaler=pickle.load(file)
# with open('model.h5', 'rb') as file:
#     model = load_model(file)
# onehot_encoder_geo=pickle.load(open('onehot_encoder_geo.pkl','rb'))
```

```python
input_data={
    'CreditScore':600,
    'Geography': 'France',
    'Gender':'Male',
    'Age':40,
    'Tenure':3,
    'Balance':60000,
    'NumOfProducts':2,
    'HasCrCard':1,
    'IsActiveMember':1,
    'EstimatedSalary':50000

}
```

```python
geo_trans=onehot_encoder_geo.transform([[input_data['Geography']]]).toarray()
# gender_trans=label_encoder_gender.transform([[input_data['Gender']]])
# geo_trans,gender_trans
geo_encoded_df=pd.DataFrame(geo_trans,columns=onehot_encoder_geo.get_feature_names_out(['Geography']))
geo_encoded_df.head()
```

```python
input_data=pd.DataFrame([input_data])
```

```python
input_df=pd.concat([input_data.reset_index(drop=True), geo_encoded_df],axis=1)
```

```python
input_df.drop('Geography',axis=1,inplace=True)
```

```python
input_df['Gender']=label_encoder_gender.transform(input_df['Gender'])
input_df.head()
```

```python
input_scaled=scaler.transform(input_df)
input_scaled
```

```python
prediction=model.predict(input_scaled)
prediction
```

```python
prediction_prob=prediction[0][0]
```

```python
if prediction_prob>0.5:
    print(f"The customer is likely to leave the bank with a probability of {prediction_prob:.2f}")
else:
    print(f"The customer is likely to stay with the bank with a probability of {1-prediction_prob:.2f}")
```
